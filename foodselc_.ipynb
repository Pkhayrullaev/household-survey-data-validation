```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

full_data = pd.read_excel(r'', sheet_name=None)

foodselc_sheet = pd.DataFrame(full_data['foodselc_'])
```

```python
Quantity = '3а Охирги давр ичида уй хўжалигингиз аъзолари томонидан сотиб олинган _${fooditemlabelc}_нинг умумий миқдорини аниқланг.'
Price_UZS = '3б Охирги давр ичида уй хўжалигингиз аъзолари томонидан сотиб олинган _${fooditemlabelc}_нинг умумий қийматини аниқланг (сўмда).'
Time_duration = '3с _${fooditemlabelc}_ни сотиб олинган даврини танланг'
Purchased_location = '4. Охирги давр ичида _${fooditemlabelc}_нинг сотиб олинган жойи:'
HouseholdID = 'm05uuid'
```

```python
items_гр          = foodselc_sheet['fooditemlabelc'].str.contains(r'\(гр\)',           case=False, na=False)
items_дона        = foodselc_sheet['fooditemlabelc'].str.contains(r'\(дона\)',         case=False, na=False)
items_ижара       = foodselc_sheet['fooditemlabelc'].str.contains('ижара хаққи',       case=False, na=False)
items_квм         = foodselc_sheet['fooditemlabelc'].str.contains(r'\(кв\.м\)',        case=False, na=False)
items_кун         = foodselc_sheet['fooditemlabelc'].str.contains(r'\(\s*кун\)', case=False, na=False)
items_литр        = foodselc_sheet['fooditemlabelc'].str.contains(r'\(литр\)',         case=False, na=False)
items_метр        = foodselc_sheet['fooditemlabelc'].str.contains(r'\(метр\)',         case=False, na=False)
items_курс        = foodselc_sheet['fooditemlabelc'].str.contains('тўлиқ курс',        case=False, na=False)
items_тўплам      = foodselc_sheet['fooditemlabelc'].str.contains(r'\(тўплам\)',       case=False, na=False)
items_хизмат      = foodselc_sheet['fooditemlabelc'].str.contains('хизмат',            case=False, na=False)

# Verify counts
all_items = (items_гр | items_дона | items_ижара | items_квм | items_кун |
             items_литр | items_метр | items_курс | items_тўплам | items_хизмат)

print(f"гр:        {items_гр.sum()}")
print(f"дона:      {items_дона.sum()}")
print(f"ижара:     {items_ижара.sum()}")
print(f"кв.м:      {items_квм.sum()}")
print(f"кун:       {items_кун.sum()}")
print(f"литр:      {items_литр.sum()}")
print(f"метр:      {items_метр.sum()}")
print(f"курс:      {items_курс.sum()}")
print(f"тўплам:    {items_тўплам.sum()}")
print(f"хизмат:    {items_хизмат.sum()}")
print(f"\nTotal covered: {all_items.sum()}")
print(f"Total rows:    {len(foodselc_sheet)}")
print(f"Uncovered:     {len(foodselc_sheet) - all_items.sum()}")
```

```python
whole_number_units = ['дона', 'тўплам', 'курс']
even_number_units  = []
fraction_ok_units  = ['гр', 'литр', 'метр', 'квм', 'кун']

min_qty_floor = {
    'гр'      : 1,
    'дона'    : 1,
    'тўплам'  : 1,
    'курс'    : 1,
    'метр'    : 0.1,
    'литр'    : 0.1,
    'квм'     : 0.1,
    'кун'     : 1,
    'ижара'   : 0,   # service-based, no floor
    'хизмат'  : 0,   # service-based, no floor
}

# Map each filtered dataframe to its unit type
unit_data = {
    'гр'      : (filtered_гр,      'гр'),
    'дона'    : (filtered_дона,    'дона'),
    'ижара'   : (filtered_ижара,   'ижара'),
    'квм'     : (filtered_квм,     'квм'),
    'кун'     : (filtered_кун,     'кун'),
    'литр'    : (filtered_литр,    'литр'),
    'метр'    : (filtered_метр,    'метр'),
    'курс'    : (filtered_курс,    'курс'),
    'тўплам'  : (filtered_тўплам,  'тўплам'),
    'хизмат'  : (filtered_хизмат,  'хизмат'),
}

purchase_locations = [
    "Бозор, павилон,Киоскадан",
    "Давлатдан",
    "Дўкондан",
    "Жисмоний шахсдан",
    "Интернет-дўконлар орқали",
    "Юридик шахсдан (ЯТТ, МЧЖ)"
]

duration_map = {
    "7 кун": 7/30,
    "1 ой" : 1,
    "3 ой" : 3,
    "6 ой" : 6,
    "1 йил": 12,
    "2 йил": 24,
    "3 йил": 36,
    "4 йил": 48,
    "5 йилдан ортиқ": 60
}

def iqr_bounds(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = max(Q1 - 1.5 * IQR, 0)
    upper = Q3 + 1.5 * IQR
    return lower, upper

def apply_iqr_flag(df, column):
    if df['fooditemlabelc'].nunique() == 1:
        lower, upper = iqr_bounds(df[column])
        return (df[column] < lower) | (df[column] > upper)
    else:
        def flag_group(group):
            lower, upper = iqr_bounds(group[column])
            return (group[column] < lower) | (group[column] > upper)
        return df.groupby('fooditemlabelc', group_keys=False).apply(flag_group)

def flag_inconsistencies_v2(df, unit_type):
    df = df.copy()
    df[Purchased_location] = df[Purchased_location].str.strip()
    df[Time_duration]      = df[Time_duration].str.strip()
    df['duration_months']  = df[Time_duration].map(duration_map)

    # CHECK 1: Zero price with store purchase
    mask_zero_price = (
        df[Purchased_location].isin(purchase_locations)
    ) & (
        df[Price_UZS] == 0
    )

    # CHECK 2: Unit price outlier (IQR per product)
    df['unit_price']        = df[Price_UZS] / df[Quantity]
    mask_unit_price_outlier = apply_iqr_flag(df, 'unit_price')

    # CHECK 3: Quantity range outlier (IQR per product)
    mask_qty_range_outlier  = apply_iqr_flag(df, Quantity)

    # CHECK 4: Monthly quantity outlier (IQR per product)
    df['qty_per_month'] = df[Quantity] / df['duration_months']
    mask_qty_outlier    = apply_iqr_flag(df, 'qty_per_month')

    # CHECK 5: Duplicate household + product
    mask_duplicate = df.duplicated(subset=[HouseholdID, 'fooditemlabelc'], keep=False)

    # CHECK 6: Round numbers
    if unit_type in fraction_ok_units:
        round_numbers = [1, 10, 100, 1000, 10000]
    else:
        round_numbers = [1000, 10000]
    mask_round = df[Quantity].isin(round_numbers) & df[Price_UZS].isin(round_numbers)

    # CHECK 7: Fraction / even number check
    if unit_type in whole_number_units:
        mask_fraction = df[Quantity] != df[Quantity].round()
    elif unit_type in even_number_units:
        mask_fraction = (df[Quantity] != df[Quantity].round()) | (df[Quantity].round() % 2 != 0)
    else:
        mask_fraction = pd.Series(False, index=df.index)

    # CHECK 8: Implausibly small quantity (absolute floor per unit)
    floor = min_qty_floor.get(unit_type, 0)
    mask_implausible_qty = df[Quantity] < floor

        # CHECK 9: Missing values in key columns
    mask_missing = (
        df[Quantity].isna() |
        df[Price_UZS].isna() |
        df[Time_duration].isna() |
        df[Purchased_location].isna()
    )

    # COMBINE FLAGS
    df['flag_zero_price']         = mask_zero_price
    df['flag_unit_price_outlier'] = mask_unit_price_outlier
    df['qty_range_outlier']       = mask_qty_range_outlier
    df['flag_qty_outlier']        = mask_qty_outlier
    df['flag_duplicate']          = mask_duplicate
    df['flag_round_numbers']      = mask_round
    df['flag_fraction_qty']       = mask_fraction
    df['flag_implausible_qty']    = mask_implausible_qty
    df['flag_missing_values'] = mask_missing

    all_flags = [
    'flag_zero_price', 'flag_unit_price_outlier', 'qty_range_outlier',
    'flag_qty_outlier', 'flag_duplicate', 'flag_round_numbers',
    'flag_fraction_qty', 'flag_implausible_qty', 'flag_missing_values'
]
    any_flag   = df[all_flags].any(axis=1)
    flagged_df = df[any_flag]

    # SUMMARY
    print(f"\n{'='*50}")
    print(f"Unit: {unit_type} | Total rows: {len(df)}")
    print(f"  flag_zero_price:          {mask_zero_price.sum()}")
    print(f"  flag_unit_price_outlier:  {mask_unit_price_outlier.sum()}")
    print(f"  qty_range_outlier:        {mask_qty_range_outlier.sum()}")
    print(f"  flag_qty_outlier:         {mask_qty_outlier.sum()}")
    print(f"  flag_duplicate:           {mask_duplicate.sum()}")
    print(f"  flag_round_numbers:       {mask_round.sum()}")
    print(f"  flag_fraction_qty:        {mask_fraction.sum()}")
    print(f"  flag_implausible_qty:     {mask_implausible_qty.sum()}")
    print(f"  Total flagged rows:       {any_flag.sum()}")
    print(f"  flag_missing_values:      {mask_missing.sum()}")

    return flagged_df, df

# ── RUN LOOP ───────────────────────────────────────────────────────────────

for unit_name, (filtered_df, unit_type) in unit_data.items():
    flagged, full = flag_inconsistencies_v2(filtered_df, unit_type)

    flagged.to_excel(f'белгиланган_{unit_name}_маҳсулотлар_C.xlsx', index=False)

    # THRESHOLDS
    # thresholds = full.groupby('fooditemlabelc')['unit_price'].apply(
    #     lambda x: pd.Series({
    #         'Q1'    : x.quantile(0.25),
    #         'Q3'    : x.quantile(0.75),
    #         'IQR'   : x.quantile(0.75) - x.quantile(0.25),
    #         'lower' : x.quantile(0.25) - 1.5 * (x.quantile(0.75) - x.quantile(0.25)),
    #         'upper' : x.quantile(0.75) + 1.5 * (x.quantile(0.75) - x.quantile(0.25)),
    #         'median': x.median(),
    #         'min'   : x.min(),
    #         'max'   : x.max()
    #     })
    # ).reset_index()

    # thresholds.to_excel(f'thresholds_C{unit_name}.xlsx', index=False)
    # print(f"  Saved: белгиланган_{unit_name}_маҳсулотлар_C.xlsx + thresholds_C{unit_name}.xlsx")
```
